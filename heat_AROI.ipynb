```python
import pandas as pd
import ipywidgets as widgets
from IPython.display import display
```

```python
# Load CSV
df = pd.read_csv("climate_data_NS_v01.csv")

# Scenario columns in your CSV
scenario_columns = {
    "Historical": "Historical median climate",
    "Mid-Century Moderate": "Mid-century median climate, moderate scenario",
    "End-Century Moderate": "End-of-century median climate, moderate scenario",
    "Mid-Century High": "Mid-century median climate, high scenario",
    "End-Century High": "End-of-century median climate, high scenario"
}

# Variables you want available
variable_codes = ["txg_hh", "txg_lp", "txgt29"]

# Create lookup from Common Name -> Variable Code
variable_rows = (
    df[df["Variable"].isin(variable_codes)]
    [["Variable", "Common Name"]]
    .drop_duplicates()
)

variable_lookup = {
    row["Common Name"]: row["Variable"]
    for _, row in variable_rows.iterrows()
}

# Widgets
region_dropdown = widgets.Dropdown(
    options=sorted(df["Region"].dropna().unique()),
    description="Region:"
)

variable_dropdown = widgets.Dropdown(
    options=sorted(variable_lookup.keys()),
    description="Variable:"
)

scenario_dropdown = widgets.Dropdown(
    options=list(scenario_columns.keys()),
    description="Scenario:"
)

output = widgets.Output()

def update_output(change=None):

    region = region_dropdown.value
    common_name = variable_dropdown.value
    scenario = scenario_dropdown.value

    variable_code = variable_lookup[common_name]
    column = scenario_columns[scenario]

    result = df[
        (df["Region"] == region) &
        (df["Variable"] == variable_code)
    ]

    output.clear_output(wait=True)

    with output:

        if result.empty:
            print("No data found.")
            return

        row = result.iloc[0]

        print(f"Region: {region}")
        print(f"Variable: {common_name}")
        print(f"Scenario: {scenario}")
        print()
        print(f"Value: {row[column]} {row['Units']}")

# Attach events
region_dropdown.observe(update_output, names="value")
variable_dropdown.observe(update_output, names="value")
scenario_dropdown.observe(update_output, names="value")

# Display widgets
display(region_dropdown)
display(variable_dropdown)
display(scenario_dropdown)
display(output)

# Initial value
update_output()
```
